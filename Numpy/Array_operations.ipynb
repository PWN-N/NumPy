# Advance Array Manipulation

## Stacking Array

```python
a = np.array([1,2,3,4])
a
```

```python
b = np.array([5,6,7,8])
b
```

### Vertical Stack

```python
np.vstack((a,b))
```

```python
np.hstack((a,b))
```

### Column Stack

```python
np.column_stack((a,b))
```

## Splitting Array

```python
c = np.arange(16).reshape(4, 4)
c
```

### Horizontal Split

```python
np.hsplit(c,2)
```

### Vertical Split

```python
np.vsplit(c,2)
```
